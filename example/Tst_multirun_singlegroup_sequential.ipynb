```python
%matplotlib tk
from os import getcwd #, chdir
startuppath = getcwd()
from mujpy.musuite import suite
from mujpy.mufit import mufit
from mujpy.mufitplot import mufitplot
```

```python
datafile = '../data/deltat_tdc_gps_0822.bin'
runlist = '822,823:834:-1' # first run first
offset = '20'
grp_calib = [{'forward':'3', 'backward':'4', 'alpha':1.13}]
```

```python
the_suite = suite(datafile, runlist , grp_calib , offset, startuppath)
dashboard_file = startuppath+'/fit/mgml.822.3-4.1_fit.json' # must exist
the_fit = mufit(the_suite,dashboard_file,scan='T')
plot_range = '0,20000,40'
the_fitplot = mufitplot(plot_range,the_fit)
```

```python
eTsTc = the_suite._the_runs_[-1][0].get_devTemperatures_vector()
```

```python
from mujpy.tools.tools import spec_prec
spec_prec(eTsTc[1])
```

```python
TsTc = the_suite._the_runs_[-1][0].get_temperatures_vector()
'{:.3f},{:.3f}'.format(TsTc[1],eTsTc[1])
```
