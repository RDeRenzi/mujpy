```python
import requests
import os
from io import BytesIO
import tarfile
url = 'http://musruser.psi.ch/cgi-bin/SearchDB.cgi'
# valori indovinati quarcando il source delle pagine web
# from Zaher http://musruser.psi.ch/cgi-bin/SearchDB.cgi?
#            &Counter=1
#            &FORMAT=MusrRoot
#            &Chk1=/psi.ch/group/lmu/public/data/gps/his/2021/deltat_tdc_gps_0822.bin,0822,GPS,2021
#            &go=TAR
# or               /psi.ch/group/lmu/public/data/lem/his/2020/lem20_his_2000.root,2000,LEM,2020
#                  /psi.ch/group/lmu/public/data/dolly/his/2019/deltat_tdc_dolly_0200.bin,0200,Dolly,2019
#                  /psi.ch/group/lmu/public/data/gps/his/2026/deltatbc_tdc_gps_2026_0200.root  ,0200,GPS,2026
# 2023 is start year for root format
rmin, rmax = 822, 834
nruns = rmax-rmin+1
form_data = {'go':'Search',
             'Rmin':str(rmin),
             'Rmax':str(rmax),
             'YEAR':'2021',
             'AREA':'GPS'}
s = requests.get(url,params=form_data)
t = s.text
st = 'Chk'
sp = '"'
dk = 29
form_data = {'go':'TAR',
             'FORMAT':'MusrRoot',
             "Counter":str(nruns)}
for k in range(nruns):
    if k==9: dk += 1
    chk = st+str(k+1)
    k1 = t.find(chk)+dk
    k2 = t.find(sp,k1)
    path = t[k1:k2]
    form_data[chk]= path
    print('{} retrieves {}'.format(chk,path))

with requests.get(url,params=form_data, stream=True) as r:
    print(r.raise_for_status())

    with tarfile.open(fileobj=r.raw,mode="r|gz") as tar:
        tar.extractall(path=os.path.join('/tmp/', 'del/'))

    
# risulta solo l'html della seconda pagina web
#print(s.headers)
#print(s.status_code)
```
