```python
import numpy as np
from matplotlib import pyplot as P
from matplotlib import animation
from ipywidgets.widgets import Output, HBox, HTML, ToggleButtons, Layout
from lorem_text import lorem
import re

# First set up the figure, the axis, and the plot element we want to animate
fig = P.figure()
ax = P.axes(xlim=(0, 2), ylim=(-2, 2))
line, = ax.plot([], [], lw=2)
# initialization function: plot the background of each frame
def init():
    line.set_data([], [])
    return line,

# animation function.  This is called sequentially
def animate(i):
    x = np.linspace(0, 2, 1000)
    y = np.sin(2 * np.pi * (x - 0.01 * i))
    line.set_data(x, y)
    return line,

tb = ToggleButtons(options = ['sequential fit','global fit'],
                                 value = 'sequential fit',
                                 tooltips = ['A1 A20 B1 B20\nsingle asymmetry fit','A21 B21 C1 C2\nmulti asymmetries fit'],
                                 layout = Layout(width='40%'))
tb.style.description_width='0%'
txt = Output(layout={'width':'50%','border':'1px solid blue'})
box = Output(layout={'width':'40%','border':'1px solid green'})
display(HBox([txt,box],layout={'width':'100%','border':'1px solid magenta'}))

def display_my_anim(anima):
    paused = False
    anim = animation.FuncAnimation(fig, anima, init_func=init,
                           frames=200, interval=20, blit=True)
    fig.canvas.toolbar_visible = False
    fig.canvas.header_visible = False
    fig.canvas.footer_visible = False
    def toggle_pause(event):
        nonlocal paused,anim
        if paused:
            anim.event_source.start() # matplotlib.__version__ >= 3.4 
            # animation.event_source.start()
        else:
            anim.event_source.stop() # if matplotlib.__version__ >= 3.4 
            # animation.event_source.stop()
        paused = not paused    
    fig.canvas.mpl_connect('button_press_event',toggle_pause)
    P.show()
    return anim
# call the animator.  blit=True means only re-draw the parts that have changed.
with txt:
    print('\n'.join(re.split('[,.?]',lorem.paragraph())))
    display(tb)
def go_now(anima,boxa):
    with boxa:
        anim = display_my_anim(anima)
    return anim
    
anim = go_now(animate,box)


# save the animation as an mp4.  This requires ffmpeg or mencoder to be
# installed.  The extra_args ensure that the x264 codec is used, so that
# the video can be embedded in html5.  You may need to adjust this for
# your system: for more information, see
# http://matplotlib.sourceforge.net/api/animation_api.html
#anim.save('animation.mp4', fps=10, extra_args=['-vcodec', 'libx264'])

#with box:
    #display(HTML("<video width='100%' height='100%' controls><source src='animation.mp4' type='libx264'></video>"))
```
