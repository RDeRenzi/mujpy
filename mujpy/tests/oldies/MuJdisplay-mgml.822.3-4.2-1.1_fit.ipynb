````python
from os import getcwd
startuppath = getcwd()
from mujpy.musuite import suite
from mujpy.mudash import dash
from mujpy.mufit import mufit
from mujpy.mufitplot import mufitplot
from ipywidgets.widgets import Output, HBox
from matplotlib.pyplot import show
datafile = '/home/roberto.derenzi/musrfit/MBT/gps/run_05_21/data/deltat_tdc_gps_0822.bin'
runlist = '822' # first run first
offset = '20'
grp_calib = [{'forward':'3', 'backward':'4', 'alpha':1.13},{'forward':'2', 'backward':'1', 'alpha':1.13}]
#
the_suite = suite(datafile, runlist , grp_calib , offset, startuppath)
dashboard_file = startuppath+'/fit/mgml.822.3-4+2-1.1_fit.json' # must exist
txt = Output(layout={'width':'45%','border':'1px solid blue'})
box = Output(layout={'width':'40%','border':'1px solid green'})
display(HBox([txt,box],layout={'width':'100%','border':'1px solid magenta'}))
with txt:
    the_fit = mufit(the_suite,dashboard_file)
plot_range = '0,20000,40'
the_fitplot = mufitplot(plot_range,the_fit, plot_out=box)
````
